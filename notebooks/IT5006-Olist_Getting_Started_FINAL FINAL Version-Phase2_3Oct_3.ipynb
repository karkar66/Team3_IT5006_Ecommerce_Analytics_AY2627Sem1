````python
# Load all Olist tables
def load_olist_data(data_dir):
    """Load all Olist CSV files into a dictionary of DataFrames."""

    tables = {
        'orders': 'olist_orders_dataset.csv',
        'order_items': 'olist_order_items_dataset.csv',
        'customers': 'olist_customers_dataset.csv',
        'products': 'olist_products_dataset.csv',
        'sellers': 'olist_sellers_dataset.csv',
        'payments': 'olist_order_payments_dataset.csv',
        'reviews': 'olist_order_reviews_dataset.csv',
        'geolocation': 'olist_geolocation_dataset.csv',
        'category_translation': 'product_category_name_translation.csv'
    }

    # Date columns to parse
    date_cols = {
        'orders': ['order_purchase_timestamp', 'order_approved_at',
                   'order_delivered_carrier_date', 'order_delivered_customer_date',
                   'order_estimated_delivery_date'],
        'order_items': ['shipping_limit_date'],
        'reviews': ['review_creation_date', 'review_answer_timestamp']
    }

    data = {}
    for name, filename in tables.items():
        filepath = os.path.join(data_dir, filename)
        parse_dates = date_cols.get(name, None)
        data[name] = pd.read_csv(filepath, parse_dates=parse_dates)
        print(f"Loaded {name}: {data[name].shape[0]:,} rows × {data[name].shape[1]} cols")

    return data

# Load data
olist = load_olist_data(DATA_DIR)
````

````python
# Quick schema overview
def schema_summary(data_dict):
    """Generate a summary of all tables."""
    summary = []
    for name, df in data_dict.items():
        summary.append({
            'Table': name,
            'Rows': f"{df.shape[0]:,}",
            'Columns': df.shape[1],
            'Memory (MB)': f"{df.memory_usage(deep=True).sum() / 1e6:.2f}",
            'Columns List': ', '.join(df.columns)
        })
    return pd.DataFrame(summary)

pd.set_option('display.max_colwidth', None)

schema_summary(olist)
````

Commentary: There are many duplicate rows for the same 5-digit zip code --> same geolocation_lat but slightly different values of geolocation_lng.

Based on research, a Brazilian zip code has eight digits in the 00000-000 format. The Olist table only provides the first five digits of the zip code. Thus, the most practical method is to assume each 5-digit zip code approximates to one corresponding latitude and longitude. In this case, our group opted to choose the median of latitude and longitude values corresponding to each 5-digit zip code. Median is chosen over average to avoid any skewness due to outlier values.

````python
# Recreate geolocation_median to ensure a clean state
geolocation_median = olist['geolocation'].groupby('geolocation_zip_code_prefix').agg(
    geolocation_lat_median=('geolocation_lat', 'median'),
    geolocation_lng_median=('geolocation_lng', 'median'),
    geolocation_city=('geolocation_city', 'first'),
    geolocation_state=('geolocation_state', 'first')
)

# Now, reset the index, dropping the old index name to avoid creating 'level_0' or 'index' columns
geolocation_median = geolocation_median.reset_index(drop=False) # drop=False explicitly makes 'geolocation_zip_code_prefix' a column

print(geolocation_median.head())
````

## 3. Merging Tables (Building the Analysis Dataset)

The Olist tables are **relational** — they connect via shared keys. Let's build a consolidated dataset for analysis.

```
orders ←→ order_items ←→ products
   ↓           ↓
customers   sellers
   ↓
reviews
```


````python
# Build consolidated analysis dataset

# Pre-calculate volume in the products table to merge directly
products_temp = products.copy()
products_temp['product_volume_cm3'] = (
    products_temp['product_length_cm'] *
    products_temp['product_height_cm'] *
    products_temp['product_width_cm']
)

# Step 1: Start with order_items (line-level)
df = olist['order_items'].copy()
print(f"Step 1 - order_items: {df.shape}")

# Step 2: Add product details (merging calculated volume instead of raw dimensions)
df = df.merge(
    products_temp[['product_id', 'product_category_name_english', 'product_weight_g', 'product_volume_cm3']],
    on='product_id',
    how='left'
)
print(f"Step 2 - + products: {df.shape}")

# Step 3: Add order details
df = df.merge(
    olist['orders'][['order_id', 'customer_id', 'order_status',
                     'order_purchase_timestamp', 'order_delivered_carrier_date', 'order_delivered_customer_date',
                     'order_estimated_delivery_date']],
    on='order_id',
    how='left'
)
print(f"Step 3 - + orders: {df.shape}")

# Step 4: Add customer details
df = df.merge(
    olist['customers'][['customer_id', 'customer_unique_id', 'customer_state', 'customer_city', 'customer_zip_code_prefix']],
    on='customer_id',
    how='left'
)
print(f"Step 4 - + customers: {df.shape}")

# Step 5: Add seller details
df = df.merge(
    olist['sellers'][['seller_id', 'seller_state', 'seller_zip_code_prefix']],
    on='seller_id',
    how='left'
)
print(f"Step 5 - + sellers: {df.shape}")

print(f"\n✅ Final consolidated dataset: {df.shape[0]:,} rows × {df.shape[1]} columns")
df.head()
````

````python
# Add derived columns for analysis
df['order_date'] = df['order_purchase_timestamp'].dt.date
df['year_month'] = df['order_purchase_timestamp'].dt.to_period('M').astype(str)
df['year'] = df['order_purchase_timestamp'].dt.year
df['month'] = df['order_purchase_timestamp'].dt.month
df['day_of_week'] = df['order_purchase_timestamp'].dt.day_name()

# Delivery metrics
df['delivery_days'] = (df['order_delivered_customer_date'] - df['order_purchase_timestamp']).dt.days
df['is_on_time'] = (df['order_delivered_customer_date'] <= df['order_estimated_delivery_date'])

# Revenue per item
df['item_revenue'] = df['price'] + df['freight_value']

print("Added derived columns:")
print(df[['order_id', 'year_month', 'day_of_week', 'delivery_days', 'is_on_time', 'item_revenue']].head())
````

#### 💡 The Root Cause

During the join operation, the diagnostic check evaluated whether the problematic 5-digit zip code prefixes existed within our reference mapping table:

> **Do the missing zip codes exist in the geolocation reference table?** $	o$ **`False`**

This key finding confirms that the specific zip code prefixes tied to these records (e.g., customer zip `38710`, seller zip `2285`) **do not exist** in the raw `olist_geolocation_dataset.csv` database.

* **Verdict:** Because these lookup keys are entirely absent from the reference catalog, a standard `left join` natively outputs `NaN` values for their coordinates and regional states.
* **Real-World Context:** This represents a classic data synchronization mismatch, typically occurring when newly established postal codes are active in transaction registers before the spatial reference index is fully updated.

### 🔍 Critical Insights & Takeaways

* **1. Regional State Tracking is Highly Standardized**
  * Both customer and seller states align near-perfectly (99%+) with the external geolocation database. State abbreviations (e.g., `SP`, `RJ`, `MG`) are standardized natively across both operational systems, making regional state-level metrics exceptionally reliable.

* **2. City-Level Discrepancies Stem from Text Variation**
  * The drop to ~90%–92% match rates on city names is primarily an artifact of spelling variation, shorthand notation, and special characters (e.g., `são paulo` vs. `sao paulo`, or handling of hyphens/spaces like `mogi-guaçu` vs `mogi guacu`).
  * Because the geolocation table coordinates were aggregated to a 5-digit zip code prefix (choosing the first alphabetically sorted city name), minor mismatches arise compared to the raw customer checkout text input.

* **3. Actionable Modeling Strategy (Last-Mile Logistics & Regression)**
  * 📌 **Key Rule:** Avoid relying on exact string matching of city names for spatial modeling or shipping distance calculations.
  * 📌 **Best Practice:** Utilize the **5-digit zip code prefix (`zip_code_prefix`)** and its mapped **latitude/longitude coordinates** directly. This bypasses typographical inconsistencies entirely and provides robust, math-friendly inputs for our machine learning pipeline.

### Exercise 1: Month-over-Month Growth Rate for Orders

The month-over-month growth rate measures how much the number of orders changed from the previous month:

$$\text{MoM growth} = \frac{\text{orders}_{m} - \text{orders}_{m-1}}{\text{orders}_{m-1}} \times 100\%$$

A positive value means orders increased, a negative value means they decreased.

Only the complete and continuous months from January 2017 to August 2018 are used. The earlier 2016 records are limited and not continuous, while September 2018 contains only a few days of data.

````python
# ============================================================
# Exercise 1: Month-over-Month Growth Rate for Orders
# ============================================================

# Step 1: Keep one row for each order
# The combined df is item-level, so the same order_id may appear
# more than once when an order contains multiple items.

monthly_order_data = df[
    ['order_id', 'order_purchase_timestamp']
].drop_duplicates(
    subset='order_id'
).copy()


# Step 2: Keep only complete and continuous months
# 2016 has limited and discontinuous records.
# September 2018 contains only a few days of data.

monthly_order_data = monthly_order_data[
    (monthly_order_data['order_purchase_timestamp'] >= '2017-01-01') &
    (monthly_order_data['order_purchase_timestamp'] < '2018-09-01')
].copy()


# Step 3: Create a year-month column

monthly_order_data['year_month'] = (
    monthly_order_data['order_purchase_timestamp']
    .dt.to_period('M')
    .astype(str)
)


# Step 4: Count the number of different orders in each month

monthly_orders = (
    monthly_order_data
    .groupby('year_month')['order_id']
    .nunique()
    .sort_index()
)


# Step 5: Calculate month-over-month growth rate

monthly_growth = monthly_orders.pct_change() * 100


# Step 6: Create a summary table

monthly_summary = pd.DataFrame({
    'Number of Orders': monthly_orders,
    'MoM Growth Rate (%)': monthly_growth
})

print("MONTH-OVER-MONTH ORDER GROWTH")
print("=" * 45)

display(monthly_summary.round(2))


# Step 7: Find the highest and lowest growth rates

highest_growth_month = monthly_growth.idxmax()
highest_growth_rate = monthly_growth.max()

lowest_growth_month = monthly_growth.idxmin()
lowest_growth_rate = monthly_growth.min()

print(
    f"Highest Growth: {highest_growth_month} "
    f"({highest_growth_rate:.1f}%)"
)

print(
    f"Largest Decline: {lowest_growth_month} "
    f"({lowest_growth_rate:.1f}%)"
)


# Step 8: Prepare the data for visualisation
# The first month is NaN because there is no previous month.

growth_plot = monthly_growth.dropna()


# Step 9: Use green for growth and red for decline

bar_colors = [
    'seagreen' if value >= 0 else 'salmon'
    for value in growth_plot
]


# Step 10: Create the bar chart

plt.figure(figsize=(12, 5))

plt.bar(
    growth_plot.index,
    growth_plot.values,
    color=bar_colors
)

# Add a horizontal line at 0%
plt.axhline(
    y=0,
    color='black',
    linewidth=1
)

plt.title(
    'Month-over-Month Order Growth Rate',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel('Month')
plt.ylabel('Growth Rate (%)')

plt.xticks(rotation=45)

plt.grid(
    axis='y',
    linestyle='--',
    alpha=0.4
)

plt.tight_layout()
plt.show()
````

### Month-over-Month Order Growth

The month-over-month growth rate compares the number of orders in each month with the previous month. Only the complete and continuous period from January 2017 to August 2018 is used, as the earlier 2016 records are limited and September 2018 contains only a few days of data.

The highest growth occurred in February 2017, when the number of orders increased by approximately 119.6% compared with January. The `price` variable does not include freight, so the result is product sales value rather than the total amount paid by customers. Since each row in the combined `df` is one item record, the prices can be summed directly by `seller_id`.

````python
# ============================================================
# Exercise 2: Top 5 Sellers by Product Sales Value
# ============================================================

# Calculate sales information for each seller
seller_summary = (
    df.groupby('seller_id', as_index=False)
      .agg(
          product_sales_value=('price', 'sum'),
          number_of_orders=('order_id', 'nunique'),
          item_records_sold=('order_id', 'size')
      )
)

# Calculate each seller's percentage of total product sales
total_product_sales = df['price'].sum()

seller_summary['share_of_total_sales'] = (
    seller_summary['product_sales_value']
    / total_product_sales
    * 100
)

# Select the top 5 sellers
top_5_sellers = (
    seller_summary
    .sort_values('product_sales_value', ascending=False)
    .head(5)
    .reset_index(drop=True)
)

# Add ranking
top_5_sellers.index = range(1, 6)
top_5_sellers.index.name = 'Rank'

# Display the results
display(
    top_5_sellers.style.format({
        'product_sales_value': 'R$ {:,.2f}',
        'number_of_orders': '{:,.0f}',
        'item_records_sold': '{:,.0f}',
        'share_of_total_sales': '{:.2f}%'
    })
)

# Combined contribution of the top 5 sellers
top_5_sales = top_5_sellers['product_sales_value'].sum()
top_5_share = top_5_sales / total_product_sales * 100

print(f"\nTop 5 Sellers' Product Sales Value: R$ {top_5_sales:,.2f}")
print(f"Top 5 Sellers' Share of Total Sales: {top_5_share:.2f}%")
````

### Exercise 3: Product Category with the Worst Delivery Performance

This analysis identifies the product category with the longest average delivery time:

$$\text{delivery days} = \text{delivered customer date} - \text{purchase date}$$

$$\overline{\text{delivery days}}_{c} = \frac{\sum \text{delivery days in category } c}{\text{number of delivered orders in category } c}$$

The category with the highest average is taken as the worst-performing. Only categories with at least 100 delivered orders are included, to avoid conclusions drawn from very small categories.

````python
# ============================================================
# Exercise 3: Worst Delivery Performance by Product Category
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt

# Minimum number of delivered orders required for comparison
MIN_ORDERS = 100

# Keep only delivered orders with a valid product category
category_delivery = df[
    (df['order_status'] == 'delivered') &
    (df['product_category_name_english'].notna())
].copy()

# Make sure the delivery columns use datetime format
date_columns = [
    'order_purchase_timestamp',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for column in date_columns:
    category_delivery[column] = pd.to_datetime(
        category_delivery[column],
        errors='coerce'
    )

# Remove records with missing delivery information
category_delivery = category_delivery.dropna(
    subset=date_columns
).copy()

# Calculate the number of days taken to deliver each order
category_delivery['delivery_days'] = (
    category_delivery['order_delivered_customer_date'] -
    category_delivery['order_purchase_timestamp']
).dt.days

# Check whether the order arrived on or before the estimated date
category_delivery['is_on_time'] = (
    category_delivery['order_delivered_customer_date'] <=
    category_delivery['order_estimated_delivery_date']
)

# Keep one row for each category-order combination
category_order_delivery = category_delivery.drop_duplicates(
    subset=['product_category_name_english', 'order_id']
).copy()

# Calculate delivery performance for each category
category_summary = (
    category_order_delivery
    .groupby('product_category_name_english', as_index=False)
    .agg(
        delivered_orders=('order_id', 'nunique'),
        average_delivery_days=('delivery_days', 'mean'),
        median_delivery_days=('delivery_days', 'median'),
        on_time_rate=('is_on_time', 'mean')
    )
)

# Change on-time rate from a decimal to a percentage
category_summary['on_time_rate'] *= 100

category_summary['late_rate'] = (
    100 - category_summary['on_time_rate']
)

# Keep categories with at least 100 delivered orders
eligible_categories = category_summary[
    category_summary['delivered_orders'] >= MIN_ORDERS
].copy()

# Rank categories by average delivery days
worst_categories = (
    eligible_categories
    .sort_values('average_delivery_days', ascending=False)
    .reset_index(drop=True)
)

# Select the 10 categories with the longest delivery time
top_10_worst = worst_categories.head(10).copy()

top_10_worst.index = range(1, len(top_10_worst) + 1)
top_10_worst.index.name = 'Rank'

# Calculate overall delivery benchmarks
overall_order_delivery = category_delivery.drop_duplicates(
    subset='order_id'
)

overall_average_days = (
    overall_order_delivery['delivery_days'].mean()
)

overall_on_time_rate = (
    overall_order_delivery['is_on_time'].mean() * 100
)

# Display the results
display(
    top_10_worst.style.format({
        'delivered_orders': '{:,.0f}',
        'average_delivery_days': '{:.2f}',
        'median_delivery_days': '{:.1f}',
        'on_time_rate': '{:.1f}%',
        'late_rate': '{:.1f}%'
    })
)

# Show the category with the worst delivery performance
worst_category = worst_categories.iloc[0]

print("\nWORST DELIVERY PERFORMANCE")
print("=" * 45)
print(
    f"Product Category: "
    f"{worst_category['product_category_name_english']}"
)
print(
    f"Delivered Orders: "
    f"{worst_category['delivered_orders']:,.0f}"
)
print(
    f"Average Delivery Days: "
    f"{worst_category['average_delivery_days']:.2f} days"
)
print(
    f"Median Delivery Days: "
    f"{worst_category['median_delivery_days']:.1f} days"
)
print(
    f"On-Time Rate: "
    f"{worst_category['on_time_rate']:.1f}%"
)
print(
    f"Overall Average Delivery Time: "
    f"{overall_average_days:.1f} days"
)
print(
    f"Overall On-Time Rate: "
    f"{overall_on_time_rate:.1f}%"
)
````

### **Product Category Sales Performance**

````python
# Import the required libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import FuncFormatter


# Select the required tables

orders = olist['orders'].copy()
order_items = olist['order_items'].copy()
products = olist['products'].copy()


# Keep only successfully delivered orders

delivered_orders = orders.loc[
    orders['order_status'] == 'delivered',
    ['order_id']
]


# Merge delivered orders with order items
# One order may contain multiple order items

category_data = delivered_orders.merge(
    order_items,
    on='order_id',
    how='inner',
    validate='one_to_many'
)


# Add product category information

category_data = category_data.merge(
    products,
    on='product_id',
    how='left',
    validate='many_to_one'
)


# Check whether the English category name already exists

if 'product_category_name_english' in category_data.columns:

    category_column = 'product_category_name_english'

else:

    # Find the category translation table automatically

    translation_tables = [
        table_name
        for table_name in olist.keys()
        if 'translation' in table_name
    ]

    if len(translation_tables) > 0:

        translation_table = olist[
            translation_tables[0]
        ]

        category_data = category_data.merge(
            translation_table,
            on='product_category_name',
            how='left',
            validate='many_to_one'
        )

        category_column = (
            'product_category_name_english'
        )

    else:

        # Use the original category name if no translation table exists

        category_column = 'product_category_name'


# Replace missing category names

category_data[category_column] = (
    category_data[category_column]
    .fillna('unknown')
)


# Calculate sales value and order volume by category

category_sales = (
    category_data
    .groupby(category_column)
    .agg(
        Sales_Value=('price', 'sum'),
        Number_of_Orders=('order_id', 'nunique')
    )
    .reset_index()
)


# Rename and format category names

category_sales = category_sales.rename(
    columns={category_column: 'Product Category'}
)

category_sales['Product Category'] = (
    category_sales['Product Category']
    .str.replace('_', ' ', regex=False)
    .str.title()
)


# Sort categories by sales value

category_sales = category_sales.sort_values(
    by='Sales_Value',
    ascending=False
)


# Select the top 10 categories

top_10_categories = category_sales.head(10).copy()


# Calculate the share contributed by the top three categories

top_3_sales = (
    category_sales
    .head(3)['Sales_Value']
    .sum()
)

total_sales = category_sales['Sales_Value'].sum()

top_3_share = top_3_sales / total_sales * 100


# Display the results

print(f'Top three category share: {top_3_share:.2f}%')

display(
    top_10_categories.style
    .hide(axis='index')
    .format({
        'Sales_Value': 'R${:,.2f}',
        'Number_of_Orders': '{:,.0f}'
    })
    .set_caption(
        'Top 10 Product Categories by Sales Value'
    )
)
````

### Delivery delay vs. review score

The idea of relating the rating to delivery performance *against the promised speed* (rather than raw delivery time) follows **Cui et al.** on the delivery-speed promise. The promise here is `order_estimated_delivery_date`; the variable is `actual − estimated` = `delay_days` ( > 0 = late ). *Full citation in the report.*

````python
# Delivery delay vs. review score — delivered orders that have a review

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


orders  = olist['orders'].copy()
reviews = olist['reviews'].copy()


# One review score per order (a few orders carry more than one review record)

review_by_order = (
    reviews
    .groupby('order_id')['review_score']
    .mean()
    .rename('review_score')
)


# Keep only delivered orders with valid delivery dates, then inner-join reviews
# (inner join -> only delivered orders that also have a review)

dr = (
    orders.loc[
        orders['order_status'] == 'delivered',
        ['order_id', 'order_delivered_customer_date', 'order_estimated_delivery_date']
    ]
    .dropna(subset=['order_delivered_customer_date', 'order_estimated_delivery_date'])
    .merge(review_by_order, on='order_id', how='inner')
)


# Delivery delay in calendar days: actual - estimated  ( > 0 means late )

dr['delay_days'] = (
    dr['order_delivered_customer_date'].dt.normalize()
    - dr['order_estimated_delivery_date'].dt.normalize()
).dt.days


print(f"delivered orders with a review : {len(dr):,}")
print(f"late rate (delay > 0)          : {(dr['delay_days'] > 0).mean():.2%}")
print()
print(dr['delay_days'].describe().round(1).to_string())
````

**Note on the tail.** `delay_days` runs out to **+188** and **−147**, but the extremes are **< 0.4%** of orders and partly a data artefact:

- **Very late (positive):** a handful of stuck / never-arrived orders were force-closed in a bulk clean-up — several share an `order_delivered_customer_date` of exactly `2017-09-19` or `2018-09-19`, so for those the "delivered" date is a bookkeeping entry, not a real delivery.
- **Very early (negative):** a few orders carry an `order_estimated_delivery_date` set ~5 months in the future (a bad estimate), which reads as a large negative delay.

Only ~345 orders are more than 30 days late (0.36%). The scatter below clips to ±30 days, where >99.6% of orders sit, so these outliers don't affect it.

````python
# Scatter: review score vs. (actual - estimated) delivery days

rng = np.random.default_rng(0)
plot = dr[dr['delay_days'].between(-30, 30)].copy()
samp = plot.sample(min(15000, len(plot)), random_state=0)
jitter = samp['review_score'] + rng.uniform(-0.18, 0.18, len(samp))   # spread the 1-5 rating so points don't overlap

fig, ax = plt.subplots(figsize=(9, 4))
ax.scatter(samp['delay_days'], jitter, s=5, alpha=0.05, color='#2E86AB')

# mean review score in 2-day bins
plot['bin'] = pd.cut(plot['delay_days'], np.arange(-30, 31, 2))
binned = plot.groupby('bin', observed=True)['review_score'].mean()
ax.plot([iv.mid for iv in binned.index], binned.values,
        color='#C73E1D', lw=2, marker='o', ms=3, label='mean score (2-day bins)')

ax.axvline(0, color='grey', ls='--', lw=1, label='delivered on the estimated date')
ax.set_xlabel('actual − estimated delivery (days)   ( > 0 = late )')
ax.set_ylabel('review_score')
ax.set_yticks([1, 2, 3, 4, 5])
ax.set_title('Review score vs. how the delivery landed against the estimate', fontweight='bold')
ax.legend()
plt.tight_layout()
plt.show()
````

### Delivery time of day vs. review score

This check follows **Kandula et al.** on delivery time-window matching — the idea that *when* in the day a parcel arrives (vs. the customer's availability) can shape satisfaction. Here the proxy is the hour of `order_delivered_customer_date`. The mean-score line below is essentially flat (`corr ≈ 0.01`), with only a slight sag for the few deliveries logged around 00:00–01:00, so **delivery time of day carries no signal for the review score** — kept as a checked-and-rejected variable. *Full citation in the report.*

````python
# Delivery hour of day vs. review score  (same population as the delay scatter above)

dr['delivery_hour'] = dr['order_delivered_customer_date'].dt.hour

print(f"delivered orders with a review    : {len(dr):,}")
print(f"corr(delivery_hour, review_score) : {dr['delivery_hour'].corr(dr['review_score']):.3f}")

rng = np.random.default_rng(0)
samp = dr.sample(min(15000, len(dr)), random_state=0)
x = samp['delivery_hour'] + rng.uniform(-0.4, 0.4, len(samp))
y = samp['review_score']   + rng.uniform(-0.18, 0.18, len(samp))

fig, ax = plt.subplots(figsize=(9, 4))
ax.scatter(x, y, s=5, alpha=0.05, color='#2E86AB')

by_hour = dr.groupby('delivery_hour')['review_score'].mean()
ax.plot(by_hour.index, by_hour.values, color='#C73E1D', lw=2, marker='o', ms=3, label='mean score by hour')
ax.axhline(dr['review_score'].mean(), color='grey', ls='--', lw=1, label='overall mean')

ax.set_xlabel('delivery hour of day (local time, 0-23)')
ax.set_ylabel('review_score')
ax.set_xticks(range(0, 24, 2))
ax.set_yticks([1, 2, 3, 4, 5])
ax.set_title('Review score vs. hour of day the order was delivered', fontweight='bold')
ax.legend()
plt.tight_layout()
plt.show()
````

Delivery time of day is therefore not a useful predictor of the rating.

## 📝On-time Delivery Rate vs. Package Weight

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Keep delivered orders with delivery information
# Keep the existing definition of is_on_time
delivered_products = df[
    (df['order_status'] == 'delivered') &
    (df['delivery_days'].notna()) &
    (df['is_on_time'].notna())
].copy()

# Ensure each item within an order is counted once
delivered_products = delivered_products.drop_duplicates(
    subset=['order_id', 'order_item_id']
)

# Sum all item weights within each order
# If any item weight is missing, leave the order total as missing
delivered_products_unique_orders = (
    delivered_products.groupby('order_id', as_index=False)
    .agg(
        total_weight_g=(
            'product_weight_g',
            lambda weights: weights.sum(min_count=len(weights))
        ),
        is_on_time=('is_on_time', 'first')
    )
)

# Exclude orders with incomplete weight information
delivered_products_unique_orders = (
    delivered_products_unique_orders.dropna(
        subset=['total_weight_g']
    ).copy()
)

# Keep the original logarithmic binning method
max_weight = delivered_products_unique_orders['total_weight_g'].max()
min_weight = delivered_products_unique_orders['total_weight_g'].min()

bins = np.unique(np.concatenate([
    np.array([0]),
    np.round(
        np.logspace(
            np.log10(min_weight + 1),
            np.log10(max_weight),
            10
        )
    )
]))

# Keep the original style of weight labels
labels = []

for i in range(len(bins) - 1):
    lower_g = bins[i]
    upper_g = bins[i + 1]

    if upper_g < 1000:
        labels.append(f'{int(lower_g)}-{int(upper_g)}g')
    else:
        labels.append(
            f'{lower_g / 1000:.1f}-{upper_g / 1000:.1f}kg'
        )

# Include the maximum weight when using right=False
# This prevents the heaviest order from being excluded
bins[-1] = np.nextafter(
    max(float(bins[-1]), float(max_weight)),
    np.inf
)

# Group orders by total product weight
delivered_products_unique_orders['weight_bin'] = pd.cut(
    delivered_products_unique_orders['total_weight_g'],
    bins=bins,
    labels=labels,
    right=False,
    include_lowest=True
)

# Calculate the on-time rate and order count for each group
on_time_rate_by_weight = (
    delivered_products_unique_orders
    .groupby('weight_bin', observed=False)
    .agg(
        is_on_time=('is_on_time', 'mean'),
        number_of_orders=('order_id', 'nunique')
    )
    .reset_index()
)

# Convert the on-time rate to a percentage
on_time_rate_by_weight['is_on_time'] *= 100

display(on_time_rate_by_weight.round(2))

# Keep the original chart style
plt.figure(figsize=(12, 7))

ax = sns.barplot(
    x='weight_bin',
    y='is_on_time',
    data=on_time_rate_by_weight,
    order=labels,
    hue='weight_bin',
    hue_order=labels,
    palette='viridis',
    dodge=False,
    legend=False
)

# Add the percentage and order count above each bar
for i, row in on_time_rate_by_weight.iterrows():

    count = int(row['number_of_orders'])
    rate = row['is_on_time']

    if count == 0:
        # An empty group has no delivery rate, rather than a 0% rate
        ax.text(
            i, 2,
            'n=0\nNo orders',
            ha='center',
            va='bottom',
            fontsize=9
        )
    else:
        ax.text(
            i, rate + 1,
            f'{rate:.1f}%\n(n={count:,})',
            ha='center',
            va='bottom',
            fontsize=9
        )

plt.title(
    'On-time Delivery Rate by Total Product Weight per Order',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Total Product Weight per Order', fontsize=12)
plt.ylabel('On-time Delivery Rate (%)', fontsize=12)

# Leave space above 100% for the labels
plt.ylim(0, 110)
plt.yticks(np.arange(0, 101, 20))

plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Explain what n represents
plt.figtext(
    0.5, 0.02,
    'Note: n = number of delivered orders in each weight group.',
    ha='center',
    fontsize=10
)

# Reserve space at the bottom for the note
plt.tight_layout(rect=[0, 0.06, 1, 1])

plt.show()
````

### On-time Delivery Rate vs. Package Volume

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Select product dimensions
product_dimensions = olist['products'][
    [
        'product_id',
        'product_length_cm',
        'product_height_cm',
        'product_width_cm'
    ]
]

# Start from item-level df, keeping the existing delivery filters
# Select only the needed columns to avoid duplicate dimension columns
delivered_products_volume = df.loc[
    (df['order_status'] == 'delivered') &
    (df['delivery_days'].notna()) &
    (df['is_on_time'].notna()),
    ['order_id', 'order_item_id', 'product_id', 'is_on_time']
].copy()

# Ensure each item within an order is counted once
delivered_products_volume = delivered_products_volume.drop_duplicates(
    subset=['order_id', 'order_item_id']
)

# Merge dimensions before aggregating orders
delivered_products_volume = delivered_products_volume.merge(
    product_dimensions,
    on='product_id',
    how='left',
    validate='many_to_one'
)

# Calculate volume for each item: length × height × width
# Missing dimensions will produce a missing item volume
delivered_products_volume['product_volume_cm3'] = (
    delivered_products_volume['product_length_cm'] *
    delivered_products_volume['product_height_cm'] *
    delivered_products_volume['product_width_cm']
)

# Sum item volumes within each order
# If any item volume is missing, leave the order total as missing
delivered_products_unique_orders_volume = (
    delivered_products_volume.groupby('order_id', as_index=False)
    .agg(
        total_volume_cm3=(
            'product_volume_cm3',
            lambda volumes: volumes.sum(min_count=len(volumes))
        ),
        is_on_time=('is_on_time', 'first')
    )
)

# Exclude orders with incomplete volume information
delivered_products_unique_orders_volume = (
    delivered_products_unique_orders_volume.dropna(
        subset=['total_volume_cm3']
    ).copy()
)

# Keep the original logarithmic binning method
max_volume = (
    delivered_products_unique_orders_volume['total_volume_cm3'].max()
)
min_volume = (
    delivered_products_unique_orders_volume['total_volume_cm3'].min()
)

# Create logarithmic bins
bins_volume = np.unique(np.concatenate([
    np.array([0]),
    np.round(
        np.logspace(
            np.log10(min_volume + 1),
            np.log10(max_volume),
            10
        )
    )
]))

# Keep the original label format
labels_volume = []

for i in range(len(bins_volume) - 1):
    lower_cm3 = bins_volume[i]
    upper_cm3 = bins_volume[i + 1]

    labels_volume.append(
        f'{int(lower_cm3)}-{int(upper_cm3)}cm³'
    )

# Include the maximum volume when using right=False
bins_volume[-1] = np.nextafter(
    max(float(bins_volume[-1]), float(max_volume)),
    np.inf
)

# Categorize orders by total product volume
delivered_products_unique_orders_volume['volume_bin'] = pd.cut(
    delivered_products_unique_orders_volume['total_volume_cm3'],
    bins=bins_volume,
    labels=labels_volume,
    right=False,
    include_lowest=True
)

# Calculate the on-time rate and order count for each volume group
on_time_rate_by_volume = (
    delivered_products_unique_orders_volume
    .groupby('volume_bin', observed=False)
    .agg(
        is_on_time=('is_on_time', 'mean'),
        number_of_orders=('order_id', 'nunique')
    )
    .reset_index()
)

# Convert the on-time rate to a percentage
on_time_rate_by_volume['is_on_time'] *= 100

display(on_time_rate_by_volume.round(2))

# Keep the original chart style and colours
plt.figure(figsize=(14, 7))

ax = sns.barplot(
    x='volume_bin',
    y='is_on_time',
    data=on_time_rate_by_volume,
    order=labels_volume,
    hue='volume_bin',
    hue_order=labels_volume,
    palette='viridis',
    dodge=False,
    legend=False
)

# Add the percentage and order count above each bar
for i, row in on_time_rate_by_volume.iterrows():

    count = int(row['number_of_orders'])
    rate = row['is_on_time']

    if count == 0:
        # Empty groups have no delivery rate
        ax.text(
            i, 2,
            'n=0\nNo orders',
            ha='center',
            va='bottom',
            fontsize=9
        )
    else:
        ax.text(
            i, rate + 1,
            f'{rate:.1f}%\n(n={count:,})',
            ha='center',
            va='bottom',
            fontsize=9
        )

plt.title(
    'On-time Delivery Rate by Total Product Volume per Order',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Total Product Volume per Order (cm³)', fontsize=12)
plt.ylabel('On-time Delivery Rate (%)', fontsize=12)

# Leave space above 100% for the labels
plt.ylim(0, 110)
plt.yticks(np.arange(0, 101, 20))

plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Explain the meaning of n
plt.figtext(
    0.5, 0.02,
    'Note: n = number of delivered orders in each volume group.',
    ha='center',
    fontsize=10
)

# Reserve space at the bottom for the note
plt.tight_layout(rect=[0, 0.06, 1, 1])

plt.show()
````

It appears that the smallest packages (0-169 cm³) have a perfect 100% on-time delivery rate. For other volume bins, the on-time delivery rate generally hovers around 90-93%, with a slight decreasing trend for the largest volume packages (129177-296208 cm³) at 88.44%. This suggests that while most packages are delivered on time, extremely large packages might face slightly more challenges in meeting delivery estimates.

````python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Ensure 'products' DataFrame includes English category names
products = olist['products'].merge(
    olist['category_translation'],
    on='product_category_name',
    how='left'
)
products['product_category_name_english'] = products['product_category_name_english'].fillna('Unknown')


# Re-construct df and add derived columns, as it might not be defined in this execution context
# Step 1: Start with order_items (line-level)
df = olist['order_items'].copy()

# Step 2: Add product details
df = df.merge(
    products[['product_id', 'product_category_name_english', 'product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']],
    on='product_id',
    how='left'
)

# Step 3: Add order details
df = df.merge(
    olist['orders'][['order_id', 'customer_id', 'order_status',
                     'order_purchase_timestamp', 'order_delivered_customer_date',
                     'order_estimated_delivery_date']],
    on='order_id',
    how='left'
)

# Step 4: Add customer details
df = df.merge(
    olist['customers'][['customer_id', 'customer_unique_id', 'customer_state', 'customer_city', 'customer_zip_code_prefix']],
    on='customer_id',
    how='left'
)

# Step 5: Add seller details
df = df.merge(
    olist['sellers'][['seller_id', 'seller_state', 'seller_zip_code_prefix']],
    on='seller_id',
    how='left'
)

# Add derived columns for analysis
df['order_date'] = df['order_purchase_timestamp'].dt.date
df['year_month'] = df['order_purchase_timestamp'].dt.to_period('M').astype(str)
df['year'] = df['order_purchase_timestamp'].dt.year
df['month'] = df['order_purchase_timestamp'].dt.month
df['day_of_week'] = df['order_purchase_timestamp'].dt.day_name()

# Delivery metrics
df['delivery_days'] = (df['order_delivered_customer_date'] - df['order_purchase_timestamp']).dt.days
df['is_on_time'] = (df['order_delivered_customer_date'] <= df['order_estimated_delivery_date'])

# Revenue per item
df['item_revenue'] = df['price'] + df['freight_value']

# Filter for delivered orders with valid delivery days
analysis_data = df[
    (df['order_status'] == 'delivered') &
    (df['delivery_days'].notna())
].copy()

# --- New: Prepare payment data ---
# Aggregate olist['payments'] to get total payment value per order
order_payments = olist['payments'].groupby('order_id')['payment_value'].sum().reset_index()
order_payments.rename(columns={'payment_value': 'total_payment_value'}, inplace=True)

# Merge total payment value with analysis_data
analysis_data = analysis_data.merge(order_payments, on='order_id', how='left')

# Drop rows where total_payment_value might be missing after merge (shouldn't happen for delivered orders)
analysis_data.dropna(subset=['total_payment_value'], inplace=True)

# --- Modified: Aggregate to order level using total_payment_value ---
# Aggregate to order level: take the first total_payment_value (already summed per order)
# and take the first delivery_days (as it's constant per order)
analysis_data_unique_orders = analysis_data.groupby('order_id').agg(
    purchase_value=('total_payment_value', 'first'), # Changed from item_revenue to total_payment_value
    delivery_days=('delivery_days', 'first') # delivery_days is consistent per order
).reset_index()

# Calculate the correlation
correlation = analysis_data_unique_orders['purchase_value'].corr(analysis_data_unique_orders['delivery_days'])
print(f"Correlation between Payment Value and Delivery Days: {correlation:.2f}")

# Visualize the relationship with a scatter plot and a regression line
plt.figure(figsize=(10, 6))
sns.regplot(
    x='purchase_value',
    y='delivery_days',
    data=analysis_data_unique_orders,
    scatter_kws={'alpha': 0.1, 's': 10},
    line_kws={'color': 'red'}
)
plt.title('Payment Value vs. Delivery Days', fontsize=16, fontweight='bold') # Updated title
plt.xlabel('Payment Value (R$)', fontsize=12) # Updated x-axis label
plt.ylabel('Delivery Days', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
````

###Analyzing carrier_to_customer_lag_exceeds_threshold vs on-time and late delivery

````python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Ensure `df` is up-to-date with necessary columns
# Calculate derived columns if they are not already present in `df`
if 'delivery_days' not in df.columns:
    df['order_date'] = df['order_purchase_timestamp'].dt.date
    df['year_month'] = df['order_purchase_timestamp'].dt.to_period('M').astype(str)
    df['year'] = df['order_purchase_timestamp'].dt.year
    df['month'] = df['order_purchase_timestamp'].dt.month
    df['day_of_week'] = df['order_purchase_timestamp'].dt.day_name()

    # Delivery metrics
    df['delivery_days'] = (df['order_delivered_customer_date'] - df['order_purchase_timestamp']).dt.days
    df['is_on_time'] = (df['order_delivered_customer_date'] <= df['order_estimated_delivery_date'])

    # Revenue per item
    df['item_revenue'] = df['price'] + df['freight_value']

# Filter for delivered orders with valid delivery days
analysis_data_delivery = df[
    (df['order_status'] == 'delivered') &
    (df['delivery_days'].notna())
].copy()

# Ensure `is_on_time` is also present at the order level (already done in df)

# Aggregate to order level to get unique order_id entries
# and ensure we have one `is_on_time` and one `delivery_days` per order.
# This also ensures `order_purchase_timestamp`, `order_delivered_customer_date`,
# and `order_delivered_carrier_date` are at the order level for lag calculation.
unique_delivered_orders = analysis_data_delivery.drop_duplicates(subset=['order_id']).copy()

# Fix: Ensure 'order_delivered_carrier_date' is available in unique_delivered_orders
# by merging it from the original orders DataFrame (olist['orders']) if it's not already present.
if 'order_delivered_carrier_date' not in unique_delivered_orders.columns:
    # Assuming 'orders' (olist['orders']) is available in the kernel state.
    # If 'orders' is not available, it would need to be loaded or passed in.
    unique_delivered_orders = unique_delivered_orders.merge(
        olist['orders'][['order_id', 'order_delivered_carrier_date']],
        on='order_id',
        how='left'
    )

# Calculate time_lag_carrier_to_customer_days for each order
unique_delivered_orders['time_lag_carrier_to_customer_days'] = (
    (unique_delivered_orders['order_delivered_customer_date'] -
     unique_delivered_orders['order_delivered_carrier_date']).dt.total_seconds() / (24 * 3600)
)

# Remove orders where this specific time lag is missing or invalid (e.g., negative)
unique_delivered_orders = unique_delivered_orders[
    (unique_delivered_orders['time_lag_carrier_to_customer_days'].notna()) &
    (unique_delivered_orders['time_lag_carrier_to_customer_days'] >= 0)
].copy()

# Calculate IQR and median for time_lag_carrier_to_customer_days
median_lag = unique_delivered_orders['time_lag_carrier_to_customer_days'].median()
Q1 = unique_delivered_orders['time_lag_carrier_to_customer_days'].quantile(0.25)
Q3 = unique_delivered_orders['time_lag_carrier_to_customer_days'].quantile(0.75)
IQR = Q3 - Q1

# Define the threshold as median + 1.5 * IQR
carrier_to_customer_lag_threshold = median_lag + 1.5 * IQR

print(f"Median carrier-to-customer lag: {median_lag:.2f} days")
print(f"IQR of carrier-to-customer lag: {IQR:.2f} days")
print(f"Calculated threshold for 'exceeds_threshold': {carrier_to_customer_lag_threshold:.2f} days")

# Create the binary feature: `carrier_to_customer_lag_exceeds_threshold`
unique_delivered_orders['carrier_to_customer_lag_exceeds_threshold'] = (
    unique_delivered_orders['time_lag_carrier_to_customer_days'] > carrier_to_customer_lag_threshold
)

# Create the `is_late` target variable
unique_delivered_orders['is_late'] = ~unique_delivered_orders['is_on_time']

# Analyze the relationship using a cross-tabulation
contingency_table = pd.crosstab(
    unique_delivered_orders['carrier_to_customer_lag_exceeds_threshold'],
    unique_delivered_orders['is_late'],
    normalize='index'
).mul(100).round(2)

print("\nContingency Table (Percentage of On-time/Late by Threshold Exceedance):")
display(contingency_table)

# Visualize the relationship with a grouped bar plot
plot_data = contingency_table.stack().reset_index(name='percentage')
plot_data.rename(columns={'is_late': 'Delivery Status'}, inplace=True)
plot_data['Delivery Status'] = plot_data['Delivery Status'].map({True: 'Late', False: 'On-Time'})
plot_data['carrier_to_customer_lag_exceeds_threshold'] = plot_data['carrier_to_customer_lag_exceeds_threshold'].map({True: 'Lag Exceeds Threshold', False: 'Lag Within Threshold'})

plt.figure(figsize=(10, 6))
sns.barplot(
    x='carrier_to_customer_lag_exceeds_threshold',
    y='percentage',
    hue='Delivery Status',
    data=plot_data,
    palette={'On-Time': 'green', 'Late': 'red'}
)
plt.title('Delivery Timeliness by Carrier-to-Customer Lag Exceeding Threshold', fontsize=14, fontweight='bold')
plt.xlabel('Carrier-to-Customer Lag', fontsize=12)
plt.ylabel('Percentage of Orders (%)', fontsize=12)
plt.ylim(0, 100)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='Delivery Status')
plt.tight_layout()
plt.show()
````

#### 📝Analysis of Carrier-to-Customer Lag Exceeding Threshold vs. Delivery Timeliness

This analysis investigates how exceeding a calculated threshold for the `carrier-to-customer` delivery time lag impacts whether an order is delivered on-time or late.

*   **Threshold Calculation:** The threshold for `carrier-to-customer_lag_exceeds_threshold` is determined using the median + 1.5 * IQR (Interquartile Range) method, identifying unusually long delivery times from the carrier to the customer.
    *   Median carrier-to-customer lag: `7.00` days
    *   IQR of carrier-to-customer lag: `7.83` days
    *   Calculated threshold: `7.00 + 1.5 * 7.83 = 18.75` days.
    Orders with a `carrier-to-customer` lag greater than `18.75` days are flagged as `True` for `carrier_to_customer_lag_exceeds_threshold`.

*   **Contingency Table Observations:**
    *   When the `carrier-to-customer` lag **does NOT exceed the threshold** (`False`):
        *   A very high percentage (`97.02%`) of these orders are delivered **on-time**.
        *   Only a small percentage (`2.98%`) are **late**.
    *   When the `carrier-to-customer` lag **DOES exceed the threshold** (`True`):
        *   The percentage of **late** deliveries dramatically increases to `69.31%`.
        *   The percentage of **on-time** deliveries drops significantly to `30.69%`.

*   **Interpretation:**
    This clearly demonstrates a strong relationship: orders that experience an unusually long `carrier-to-customer` lag (exceeding the calculated threshold) are far more likely to be delivered late. This feature is a powerful predictor for late delivery and highlights the critical importance of efficient last-mile logistics. It should be a highly valuable feature for a classification model aiming to predict on-time vs. late delivery.

*   **`price`** (or `total_order_price`) / **`total_payment_value`**: Although correlations with raw delivery days were weak, in a classification context, high-value orders might be subject to different, possibly more stringent, handling procedures which could affect their on-time status.
*   **`number_of_items_in_order`**: Orders with many distinct items might require more complex fulfillment, leading to longer processing times and a higher chance of missing delivery windows.

**5. Geographical Features:**
*   **`customer_state`** / **`seller_state`**: Delivery performance (including on-time rates) was shown to vary significantly across different customer and seller states, and for intra-state vs. inter-state routes. Certain state-to-state routes might have inherent logistical challenges leading to more frequent delays.
*   **`customer_lat`**, **`customer_lng`**, **`seller_lat`**, **`seller_lng`**: These can be used to calculate a more precise **`distance_between_customer_seller`**, a crucial factor that directly impacts transit time and thus the probability of on-time delivery.

**6. Operational Features (Derived/Aggregated):**
*   **`is_weekend_purchase`**: A boolean flag (derived from `day_of_week`) explicitly capturing whether an order was placed on a weekend, as weekend orders were observed to have longer delivery times.
*   **`carrier_to_customer_lag_exceeds_threshold`**: A binary feature indicating if the `time_lag_carrier_to_customer` exceeds a predefined threshold (e.g., median + 1.5 IQR). This feature demonstrated a very strong relationship with late delivery, with orders exceeding the threshold being significantly more likely to be late. This makes it a highly valuable predictor for a classification model.

NOTE: When making a prediction at the time an order is purchased, we cannot use features that rely on information that becomes available only after the purchase has been made or after parts of the delivery process have started. From the previously discussed list for building a classification model to predict on-time vs. late delivery, the following features would not be available at the time of purchase:

`time_lag_purchase_to_approval_days`: This measures the duration between purchase
and approval, and thus the approval event must have occurred for this value to be known.
`time_lag_approval_to_carrier_days`: This measures the duration between order approval and the handover to the carrier. Both of these events occur after the initial purchase.
`carrier_to_customer_lag_exceeds_threshold`: This binary feature depends on the time_lag_carrier_to_customer, which is calculated using `order_delivered_customer_date` and `order_delivered_carrier_date`. Both the carrier handover and customer delivery events happen significantly after the purchase.

However, a dynamic version of `carrier_to_customer_lag_exceeds_threshold` can be very valuable for ongoing monitoring of outstanding orders and flagging at-risk deliveries! Here's how it would work:

Training the Model: In the training dataset, we would use the actual `carrier_to_customer_lag_exceeds_threshold` (based on when the order was actually delivered to the customer). The model learns the relationship between exceeding this threshold and the final on-time/late outcome.

Real-time Monitoring of Outstanding Orders: For an order that has been placed but not yet delivered (an 'outstanding order'):

If `order_delivered_carrier_date` has occurred: We can calculate the current duration since carrier handover: (`current_timestamp` - `order_delivered_carrier_date`). Let's call this `current_carrier_to_customer_lag_days`.
We would then compare this current_carrier_to_customer_lag_days against the `carrier_to_customer_lag_threshold` (which was derived from the training data). If `current_carrier_to_customer_lag_days` > `carrier_to_customer_lag_threshold`, we can set a dynamic `at_risk_flag` to True for that order.
This `at_risk_flag` (or a similar dynamic feature) essentially provides a real-time signal to your model, indicating that an outstanding order is now exhibiting a prolonged `carrier-to-customer` lag that, based on historical data, strongly correlates with late delivery. So, while the final `carrier_to_customer_lag_exceeds_threshold` isn't known at the moment of purchase, an evolving indicator based on the same logic can indeed be a powerful feature for identifying at-risk orders as they progress through the delivery pipeline.

### Summary of Data Preprocessing & Cleaning Methodology to generate the smartcommerce_consolidated.csv dataset

This section outlines the exact data cleaning, relational integration, and feature engineering steps applied to the Olist source tables to construct the saved consolidated analysis dataset (`smartcommerce_consolidated.csv`).

---

#### 1. Data Ingestion & Automatic Datetime Parsing
* **Temporal Formatting:** During ingestion, raw date columns from the source tables were parsed into native Pandas `datetime64[ns]` format. This standardized handling for downstream date calculations:
  * *Orders Table:* `order_purchase_timestamp`, `order_approved_at`, `order_delivered_carrier_date`, `order_delivered_customer_date`, `order_estimated_delivery_date`
  * *Order Items Table:* `shipping_limit_date`
  * *Reviews Table:* `review_creation_date`, `review_answer_timestamp`
* **Error Coercion:** Missing or malformed dates were safely standardized using `errors='coerce'` to prevent parsing crashes.

#### 2. Relational Merging & Schema Denormalization
To build a cohesive single table for analysis, we denormalized the core transactional tables using a sequential left-join chain centered on the line-item level table:
* **Base Table:** Started with `order_items` (112,650 rows).
* **Product Catalog:** Left-joined with the `products` table via `product_id` to attach product category, weight, and dimensions. Rather than keeping the raw dimensions (`product_length_cm`, `product_height_cm`, and `product_width_cm`), we calculated product volume first and kept only the engineered column.
* **Order Transactions:** Left-joined with the `orders` table via `order_id` to map status and critical delivery timestamps.
* **Customer Registry:** Left-joined with the `customers` table via `customer_id` to fetch customer unique identifiers, city, state, and zip code prefixes.
* **Seller Registry:** Left-joined with the `sellers` table via `seller_id` to align seller states and zip code prefixes.
* **Final Shape:** The resulting merged dataset contains exactly **112,650 rows** and **38 columns** (after adding the custom geolocation coordinates and dropping the raw physical dimensions).

#### 3. Category Name Localization & Imputation
* **English Translation:** Integrated the original Portuguese category names (`product_category_name`) with the translation lookup dictionary table (`category_translation`) via a left-join to map English category names.
* **Missing Class Imputation:** Missing category assignments (610 products with no original category, and 13 products with missing English translations) were explicitly imputed with the string placeholder `'Unknown'` to retain them in the dataset safely.

#### 4. Spatial Geolocation Aggregation & Calibration
* **Deduplication:** The raw `geolocation` dataset contained extensive duplicate entries and tiny coordinate variances for identical zip code prefixes.
* **Outlier-Resilient Medians:** Grouped coordinate entries by the 5-digit `geolocation_zip_code_prefix` and calculated the **median** latitude and longitude (rather than the mean) to shield the geospatial averages from skewing outliers.
* **Bilateral Mapping:** Successfully matched and joined the calibrated coordinates back to the transaction DataFrame twice:
  * Once for the buyer (`customer_zip_code_prefix` $\to$ `customer_lat`, `customer_lng` along with median city and state)
  * Once for the seller (`seller_zip_code_prefix` $\to$ `seller_lat`, `seller_lng` along with median city and state)

#### 5. Base Engineered Features
To simplify business and operational reporting, several derived indicators were computed directly inside the consolidated output dataset:
* **Date Components:** Extracted `order_date` (as date object), `year_month` (string representation, e.g., `'2017-09'`), `year`, `month`, and `day_of_week` from the transactional purchase timestamp.
* **Product Volume:** Created `product_volume_cm3` by multiplying the three spatial dimensions, then excluded the original spatial components to minimize redundant columns.
* **Continuous Delivery Lead Time:** Calculated `delivery_days` as the integer days difference between customer delivery and order purchase:
  $$\text{delivery\_days} = \text{order\_delivered\_customer\_date} - \text{order\_purchase\_timestamp}$$
* **On-Time Flag:** Created a boolean variable `is_on_time` comparing customer delivery against expectations:
  $$\text{is\_on\_time} = (\text{order\_delivered\_customer\_date} \le \text{order\_estimated\_delivery\_date})$$
* **Gross Item Revenue:** Calculated `item_revenue` as the total transaction cost per item:
  $$\text{item\_revenue} = \text{price} + \text{freight\_value}$$

  * *Time-Series Trends* (monthly order volumes and temporal purchase spikes).
